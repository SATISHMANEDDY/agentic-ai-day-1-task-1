```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

system_prompt = SystemMessage(
    content="""
    You are VVS AI, a personal AI assistant.

    Your creator is Satish.

    Your personality:
    - Friendly and intelligent
    - Speak in simple English and Tenglish
    - Help with Python, AWS, AI and studies
    - Explain everything step by step
    - Remember previous messages in this conversation
    """
)

print("VVS AI Personality Created!")
```

```python
chat_history = [system_prompt]

print("Welcome to VVS AI!")
print("Type 'exit' to stop.")

while True:
    question = input("\nYou: ")

    if question.lower().strip() == "exit":
        print("VVS AI: Goodbye!")
        break

    chat_history.append(HumanMessage(content=question))

    try:
        response = llm.invoke(chat_history)

        chat_history.append(
            AIMessage(content=response.content)
        )

        print("\nVVS AI:", response.content)

    except Exception as e:
        chat_history.pop()
        print("Error:", e)
```

```python
import gradio as gr

from langchain_core.messages import (
    SystemMessage,
    HumanMessage,
    AIMessage
)

def vvs_chat(message, history):

    messages = [system_prompt]

    for chat in history:
        if chat["role"] == "user":
            messages.append(
                HumanMessage(content=chat["content"])
            )

        elif chat["role"] == "assistant":
            messages.append(
                AIMessage(content=chat["content"])
            )

    messages.append(HumanMessage(content=message))

    response = llm.invoke(messages)

    return response.content

print("VVS AI Chat Function Ready!")
```

```python
demo = gr.ChatInterface(
    fn=vvs_chat,
    title="VVS AI - Personal Assistant",
    description="Your personal AI assistant powered by Groq and GPT-OSS.",
    examples=[
        "Hello VVS AI!",
        "Explain Python in marati",
        "What is AWS EC2?",
        "Help me learn Machine Learning",
        "Write a Python program"
    ]
)

demo.launch(share=True)
```
